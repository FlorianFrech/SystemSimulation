# Modelica

[Modelica Tutorial](https://mbe.modelica.university/behavior/equations/electrical/)

## What is Modelica?

 - Modelica is a high-level declarative language for describing mathematical behavior.
- It is typically applied to engineering systems and can be used to easily describe the behavior of different types of engineering components (e.g., springs, resistors, clutches, etc.)
 - These components can then be combined into subsystems, systems or even architectures.

## Why Modelica?

- Modelica compilers allow engineers to focus on high-level mathematical descriptions of component behavior and get high performance simulation capability without having to be deeply knowledgeable about complex topics like
     - differential-algebraic equations,
     - symbolic manipulation,
     - numeric solvers,
     - code generation,
     - post-processing, etc..

- wide range of modeling formalisms that allow the description of both **continuous and discrete behavior framed in the context of hybrid differential-algebraic equations**
- supports both **causal** (often used for control system design) and **acausal** (often used in creating schematic oriented physical designs) approaches within the same model

- designed from the start as an open language
- the specification is freely available and tool vendors are encouraged to support the import and export of Modelica

## What Modelica allows me to do?

 - modeling the behavior of engineering systems in nearly any engineering domain
 - supports both physical design and control design in a single language
 - multi-domain: so it doesn’t impose any artificial boundaries that restrict its use to select engineering domains or systems
 - provides a complete set of capabilities for building lumped system models of nearly any engineering system

### Initialization

**Overview**
 - behavior is represented by the equations contained in a model as well as the initial conditions given to the state variables in the model
 - initial conditions are computed by combining the normal equations (present in equation sections) with any initial equations (present in initial equation sections)
 - Question: How many initial conditions are required? In order to have a well-posed initialization problem, we need to have the same number of equations in the initial equation sections as we have states in our system
 - we can get away with having too few, because tools can augment the initial equations we provide with additional ones until the problem is well-posed
 - we may not be able to solve a problem where we have too many initial equations (since this depends on the tools ability to recognize and eliminate redundant equations and different tools provide different levels of support for this)
 - states in each of our examples so far are the variables that appear inside the der(...) operator. (variable that we differentiated in those examples is a state)

**ODEs**
 - it will not always be the case that every variable that we differentiate will be a state
 - When dealing with ODEs, every differentiated variable is a state, which, in turn, means that you need an initial equation for each of these differentiated variables
 - in subsequent chapters we will eventually run across examples that are so-called differential-algebraic equations (DAEs)
    - only some of the differentiated variables can be considered states
 - Modelica tools perform something called “index reduction”
    - Index reduction transforms the DAEs into ODEs: compilers will transform whatever DAE problem contained in our Modelica code into this relatively easy to explain ODE form
