## Contact Modeling using FEM

### 1. Hertzian Contact Model

[Contact mechanics](https://en.wikipedia.org/wiki/Contact_mechanics) is the study of the deformation of solids that touch each other at one or more points. The [Hertzian contact model](https://en.wikipedia.org/wiki/Hertzian_contact_stress) describes the stress and deformation that occurs when two curved surfaces come into contact.

**Assumptions in Hertzian Theory:**
- The strains are small and within the elastic limit of the materials.
- The surfaces are continuous and non-conforming, meaning they only touch at a point or along a line.
- Each body can be considered as an elastic half-space.
- The surfaces are frictionless.

**Contact between a spehere and a half-space:**

When a sphere of radius $R$ is pressed against a flat surface with a force $F$, the contact area is circular with radius $a$.

![Hertzian Contact Model](HertzianContact.jpg)

An elastic sphere of radius $R$ indents an elastic half-space where total deformation is $d$, causing a contact area of radius

$$a={\sqrt {Rd}}$$

The applied force $F$ is related to the indentation depth $d$ by

$$F = \frac{4}{3} E^* R^{\tfrac{1}{2}} d^{\tfrac{3}{2}}$$

where $E^*$ is the effective modulus of elasticity given by

$$\frac{1}{E^*} = \frac{1 - \nu_1^2}{E_1} + \frac{1 - \nu_2^2}{E_2}$$

for two materials with Young's moduli $E_1$, $E_2$ and Poisson's ratios $\nu_1$, $\nu_2$.

The distribution of pressure $p(r)$ within the contact area is given by

$$p(r) = p_0 \left(1 - \frac{r^2}{a^2}\right)^{\tfrac{1}{2}}$$

where $p_0$ is the maximum contact pressure at the center of the contact area, and $r$ is the radial distance from the center.

The maximum contact pressure $p_0$ is given by

$$p_0 = \frac{3F}{2\pi a^2}$$

The radius of the circle is related to the applied load $F$ by

$$a^3 = \frac{3FR}{4E^*}$$

The total deformation $d$ is related to the applied load $F$ by

$$d = \left(\frac{9F^2}{16E^{*2}R}\right)^{\tfrac{1}{3}}$$

**Example Calculation:**

We consider the impact of the pendulum head with a spherical head of radius $R = 0.05 m$ on a flat wall. The pendulum head hits the wall when the pendulum angle is $\theta = 0$, means the pendulum is in the lowest position. The velocity of the pendulum head at this position can be calculated using energy conservation.
