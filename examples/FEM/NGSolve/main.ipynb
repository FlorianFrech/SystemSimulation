### FEM Pendulum Configuration

```python
from fem.config.config import GeometryParameters
geom_params = GeometryParameters()
print(geom_params)
```

```python
from fem.config.config import ContactParameters
contact_params = ContactParameters()
print(contact_params)
```

```python
from fem.config.config import MaterialParameters
material_params = MaterialParameters()
print(material_params)
```

```python
from fem.config.config import InitialConditionParameters
init_cond_params = InitialConditionParameters()
init_cond_params.angular_position_deg = 30
init_cond_params.angular_velocity     = -1
init_cond_params.angular_acceleration = -0.5
print(init_cond_params)
```

```python
from fem.config.config import SimulationParameters
sim_params = SimulationParameters()
sim_params.t_start = 0.0
sim_params.t_end   = 10.0
print(sim_params)
```

### Pendulum Instantiation

```python
from fem.pendulum import FEMPendulum
myPendulum = FEMPendulum(geom_params=geom_params,
                         mat_params=material_params,
                         init_params=init_cond_params,
                         contact_params=contact_params)

myPendulum.initialize()
myPendulum._visualize_initial_conditions()
```

```python
myPendulum.initialize_scene()
```

```python
myPendulum.step(0, 0.1)
```

```python
import numpy as np
out = myPendulum.get_outputs()
q_state_deg = np.rad2deg(out['q'])
print(f"Angular Position (deg):   {q_state_deg}")
print(f"Angular Velocity (rad/s): {out['omega']}")
```

```python
myPendulum.step(0.1, 0.2)
```
