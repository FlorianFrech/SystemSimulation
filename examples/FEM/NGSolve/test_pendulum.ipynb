```python
from elastic_pendulum_impact import *

# Create pendulum simulation
pendulum = ElasticPendulumContact(
    pendulum_type=PendulumType.PLANAR_2D,
    geometry_params=GeometryParameters(
        pendulum_head_radius=0.075,
        pendulum_length=0.4,
        pendulum_rod_radius=0.02,
        wall_angle_deg=90.0,
        wall_height=0.4,
        wall_thickness=0.05,
        wall_width=0.05
    ),
    material_params=MaterialParameters(
        youngs_modulus=21000.0,
        poisson_ratio=0.2,
        density=7000.0,
        material_model=MaterialModel.NEO_HOOKE
    ),
    mesh_params=MeshParameters(
        max_element_size=0.025,
        mesh_order=3
    )
)

pendulum.create_geometry_and_mesh()
pendulum.show_geometry()

initial_conditions = InitialConditions(angular_position_deg=-180.0)
pendulum.set_initial_conditions(initial_conditions)
pendulum.show_current_state()
```

```python
# Add this after mesh creation to see available boundaries
print("Available boundaries:", pendulum._mesh.GetBoundaries())
print("Available materials:", pendulum._mesh.GetMaterials())
```

```python
pendulum.setup_simulation(SimulationParameters(
     end_time=0.1,
     time_step=0.01
       ))

pendulum.simulate(verbose=False)
```

```python
animation_settings = AnimationSettings(
     speed_multiplier=2.0,
     show_deformation=True,
     show_velocity_vectors=True
 )
pendulum.animate_results(animation_settings)

print(f"DOFs: {pendulum.num_degrees_of_freedom}")
print(f"Time span: {pendulum.simulation_time_span}")
```
