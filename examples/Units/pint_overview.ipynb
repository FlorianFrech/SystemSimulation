# Pint Overview


## Unit Registry

```python
from pint import UnitRegistry
ureg = UnitRegistry()
```

```python
L = 3 * ureg.meter              # Quantity
v = 4 * ureg.meter / ureg.second
a = v / (2 * ureg.second)
L_magnitude = L.magnitude       # 3
L_units = L.units               # meter
L_in_ft = L.to('feet')          # 9.8425... ft

print(f"Length: {L}, Magnitude: {L_magnitude}, Units: {L_units}, in feet: {L_in_ft}")
```

```python
ureg = UnitRegistry()

# Create a quantity using the registry
speed = 10 * ureg.meter / ureg.second
print(f"Speed: {speed}")
print(f"Speed units: {speed.units}")
print(f"Speed dimensionality: {speed.dimensionality}")
```

```python
# Pint can format units in compact (abbreviated) form
print("Default format:", speed)
print("Compact format:", f"{speed:~}")   # ~ modifier for compact units
print("Pretty format: ", f"{speed:P}")   # P modifier for pretty printing
```

## Unit

```python
from pint import Unit
```

```python
d = 10 * ureg.cm
d.to('m')                 # 0.1 meter
d.ito('m')                # in-place: d becomes 0.1 meter

E = 15 * ureg.kJ
E.to_base_units()         # -> 15000 kilogram * meter ** 2 / second ** 2
E.to_compact()            # -> 15 kilojoule  (nice human-friendly scaling)

print(f"Distance: {d}, Energy: {E}")
print(f"Distance in meters: {d.to('m')}\nEnergy in base units: {E.to_base_units()}\nEnergy compact: {E.to_compact()}")
```

```python
# Print short version of the unit
print(f"Short version of the unit: {ureg.meter:~}")

# Print pretty version of the unit
print(f"Pretty version of the unit: {ureg.meter:P}")

```

## Quantity

```python
from pint import Quantity as Q
```

```python
speed = Q(10, 'meter / second')

# Pint can format units in compact (abbreviated) form
print("Default format:", speed)
print("Compact format:", f"{speed:~}")  # ~ modifier for compact units
print("Pretty format:", f"{speed:P}")   # P modifier for pretty printing
```

```python
V = ureg.volt
ohm = ureg.ohm

v = 1.2 * ureg.m / (250 * ureg.ms)     # 4.8 m/s
I = (5 * V) / (2 * ohm)                 # 2.5 A
Xc = 1 / (2 * ureg.pi * 50 * ureg.hertz * 100 * ureg.uF)  # capacitive reactance

print(f"speed: {v}\nCurrent: {I}\nCapacitive Reactance: {Xc}")
```

## Dimensional Analysis

```python
from pint.errors import DimensionalityError

speed = 10 * ureg.meter / ureg.second

mass = 5 * ureg.kilogram

mass_dim = mass.dimensionality
seed_dim = speed.dimensionality

print(mass_dim, mass.unit_items())
print(seed_dim, speed.unit_items())


try:
    result = mass + speed
except DimensionalityError as e:
    print(f"DimensionalityError: {e}")
```

```python
q_SI = (10 * ureg.kJ).to_base_units()  # kg·m²/s²
val_SI, unit_SI = q_SI.magnitude, str(q_SI.units)  # 10000.0, 'kilogram * meter ** 2 / second ** 2'

print(f"SI Value: {val_SI}, SI Unit: {unit_SI}")
```

```python
print(f"{q_SI:~}")  # Compact form
```

```python
q = 12345 * ureg('ohm')
print(f"{q:~P}")
q.to_compact() 
```

## Compatibility Checks

```python
(5*ureg.cm).is_compatible_with(ureg.m)      # True
```

```python
(5*ureg.cm).is_compatible_with(ureg.s)      # False
```

```python
(3*ureg.V).check('[mass]')                  # raises in recent Pint versions
```

```python
(3*ureg.V).dimensionality == ureg('[mass]').dimensionality  # False
```
