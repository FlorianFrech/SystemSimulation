# Pint Overview

Python library for handling physical quantities with units.

````python
from pint import UnitRegistry

ureg = UnitRegistry()
````

````python
m = ureg("meter")
````

````python
dt = 1 * ureg.s
ds = 1 * ureg.m
dv = ds / dt
print(f"{ds} / {dt} = {dv}")
````

````python
print("Default         :", dv)
print("Compact         :", f"{dv:~}")
print("Pretty          :", f"{dv:P}")
print("Compact & Pretty:", f"{dv:~P}")
````

````python
ureg.formatter.default_format = "~P"
````

````python
ds_vec = [1, 2, 3] * ureg.m
dv_vec = ds_vec / dt
print(f"{ds_vec} / {dt} = {dv_vec}")
````

````python
dv_vec

# Get unit of variable
print(dv_vec.magnitude)
````

````python
v = 100 * ureg.km / ureg.h

print(f"v = {v}")
print(f"v in base units = {v.to_base_units()}")
print(f"v in reduced units = {v.to_reduced_units()}")
````

````python
print("Dimensionality of ds:", ds.dimensionality)
print("Dimensionality of dt:", dt.dimensionality)
print("Dimensionality of dv:", dv.dimensionality)
````

````python
d = 10 * ureg.cm

d.ito("m")  # in-place: d becomes 0.1 meter

E = 15 * ureg.kJ

print(f"Distance: {d}, Energy: {E}")
print(
    f"Distance in meters: {d.to('m')}\nEnergy in base units: {E.to_base_units()}\nEnergy compact: {E.to_compact()}"
)
````

**Unit Formatting**

````python
# Print short version of the unit
print(f"Short version of the unit: {ureg.meter:~}")

# Print pretty version of the unit
print(f"Pretty version of the unit: {ureg.meter:P}")

# Dimensionality checks
print((ds / dt).is_compatible_with(dv))
print((ds / dt).dimensionality == dv.dimensionality)
````

**Dimensional Analysis**

````python
from pint.errors import DimensionalityError

speed = 10 * ureg.meter / ureg.second

mass = 5 * ureg.kilogram

mass_dim = mass.dimensionality
seed_dim = speed.dimensionality

print(mass_dim, mass.unit_items())
print(seed_dim, speed.unit_items())


try:
    result = mass + speed
except DimensionalityError as e:
    print(f"DimensionalityError: {e}")
````

**Quantity**

````python
from pint import Quantity as Q
````

````python
speed = Q(10, "meter / second")

# Pint can format units in compact (abbreviated) form
print("Default format:", speed)
print("Compact format:", f"{speed:~}")  # ~ modifier for compact units
print("Pretty format:", f"{speed:P}")  # P modifier for pretty printing
````

**Dimensionality and Unit Items**

````python
force = Q(20, "newton")
force.ito_base_units()  # Convert to base units in-place
print("Force in base units:", force)
print("Dimensionality of force:", force.dimensionality)
print("Unit items of force:", force.unit_items())
````

````python
work = Q(50, "joule")
work.ito_base_units()  # Convert to base units in-place
print("Work in base units:", work)
print("Dimensionality of work:", work.dimensionality)
print("Unit items of work:", work.unit_items())
````

````python
energy = Q(100, "kilojoule")
energy.ito_base_units()  # Convert to base units in-place
print("Energy in base units:", energy)
print("Dimensionality of energy:", energy.dimensionality)
print("Unit items of energy:", energy.unit_items())
````

````python
power = Q(200, "watt")
power.ito_base_units()  # Convert to base units in-place
print("Power in base units:", power)
print("Dimensionality of power:", power.dimensionality)
print("Unit items of power:", power.unit_items())
````

### Usage for Framework

 - Allows to define and manipulate physical quantities with units
 - Supports unit conversions and dimensional analysis for consistency checks
 - Integrates with NumPy for array operations with units
 - Lightweight wrapper allows unit cheks before transferring data to OpenModelica backend
 - Unit checks shall detect modeling errors early in the development process
 - Pint provides all required functionality, no need to implement custom solution, only a lightweight wrapper around Pint is needed

Source: [Pint Documentation](https://pint.readthedocs.io/en/stable/)
