**Equation-based Model Definition**

```python
import sys
from pathlib import Path
repo_root = Path.cwd().parent.parent
sys.path.insert(0, str(repo_root))
```

```python
from SysSimX.dsl.model import Model
from SysSimX.dsl.block_interfaces import RealInput, RealOutput
from SysSimX.dsl.variable import Variable
from SysSimX.dsl.equation import Equation
from SysSimX.dsl.math_utilities import sin, pi, time
```

**Create a new System Model**

```python
FilterSystem = SystemModel("FilterSystem")
```

```python
def create_model_directory(model_name: str):
    """
    Create a directory for the model if it does not exist.
    """
    model_path = Path.cwd().parent.parent / "Models" / model_name
    model_path.mkdir(parents=True, exist_ok=True)
    return model_path
```

```python
system_model_name = "FilterSystem"

model_path =create_model_directory(system_model_name)
```

```python
# Instantiate the Model class
signal_source = Model("SignalSource")

# Define the package
package_name = "FilterSystem"
signal_source.within = package_name

# Add outputs
output = RealOutput("Uin", "V")
signal_source.add_output(output)

# Define parameters
param_A1 = Variable("A1", "Real", "V", 60)
param_A2 = Variable("A2", "Real", "V", 5)
param_f1 = Variable("f1", "Real", "Hz", 0.159)
param_f2 = Variable("f2", "Real", "Hz", 7.96)

# Add parameters to the model
parameters = [param_A1, param_A2, param_f1, param_f2]
for param in parameters:
    signal_source.add_parameter(param)

# Define internal variables
internal_var_s1 = Variable("s1", "Real", "V")
internal_var_s2 = Variable("s2", "Real", "V")
signal_source.add_variable(internal_var_s1)
signal_source.add_variable(internal_var_s2)

# Define equations
eqn1_lhs = internal_var_s1
eqn1_rhs = param_A2 * sin(2 * pi * param_f1 * time)
eqn1 = Equation(eqn1_lhs, eqn1_rhs)
signal_source.add_equation(eqn1)

eqn2_lhs = internal_var_s2
eqn2_rhs = param_A1 * sin(2 * pi * param_f2 * time)
eqn2 = Equation(eqn2_lhs, eqn2_rhs)
signal_source.add_equation(eqn2)

signal_source.to_mo()
print(signal_source.modelica_code)
```

```python
signal_source.save(model_path / f"{signal_source.name}.mo")
```
