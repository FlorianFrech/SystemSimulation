## General Mechanical Variational Problem for Solid Mechanics

### Strong, weak form and variational structure

The strong form of the general mechanical problem is given by the static balance of linear momentum along with Dirichlet and Neumann boundary conditions.

$$\begin{align}
\sigma_{ij,j} + \rho b_i = 0 \quad &\text{in } \Omega \\
u_i = \hat u_i \quad &\text{on } \partial \Omega_D \\
\sigma_{ij} n_j = \hat t_i \quad &\text{on } \partial \Omega_N
\end{align}$$

- Case of linear elasticity: $\sigma_{ij} = \mathbb C_{ijkl} \epsilon_{kl} = \mathbb C_{ijkl} \epsilon_{k,l}$, we notice that the linear momentum balance reduces to
$$
(\mathbb C_{ijkl} u_{k,l})_{,j} + \rho b_i = 0
$$
- The classical solution must satisfy $u_i \in C^2(\Omega)$

- This strong form has a variational structure. We start from the total potential energy (total energy - work done by external forces)
$$
\mathcal{I}[u] = \underbrace{\int_\Omega W(\epsilon(u)) \, dV}_{\text{internal stored energy}}
-
\underbrace{\int_\Omega \rho \mathbf b \cdot  u \, dV + \int_{\partial \Omega_N} \mathbf{\hat t} \cdot \mathbf u \, dS}_{\text{work done by external forces}}
$$
- We assume that $\rho \mathbf b$ and $\hat t$ are independent of $u$.
- The strain energy density is defined as
$$
W = W (\epsilon) \qquad \text{s.t.} \qquad \sigma_{ij} = \frac{\partial W}{\partial \epsilon_{ij}}
$$
- For the special case of linear elasticity, the strain energy density is given by
$$
W(\epsilon) = \frac{1}{2} \mathbb C_{ijkl} \epsilon_{ij} \epsilon_{kl} = \frac{1}{2} \mathbb C_{ijkl} u_{i,j} u_{k,l}
$$
- The energy functional requires $u_i \in \mathcal{U} = \{ u_ i \in H^1(\Omega): u_i = \hat u_i \text{ on } \partial \Omega_D \}$ for each displacement component $i$.

- Equivalence between solving the strong form and minimizing the energy functional can be shown.
- The energy functional is equivalent to the principle of virtual work.
- Application of the divergence theorem gives:
$$
\delta \mathcal{I} = 0 =
- \int_\Omega (\sigma_{ij, j} + \rho b_i) \delta u_i \, dV
+ \int_{\partial \Omega_N} (\sigma_{ij} n_j - \hat t_i) \delta u_i \, dS
$$
- This must hold for all admissible variation $\delta \mathbf u$ wuch that the terms in parentheses must vanish
- This is equivalent to the strong form of the problem. Hence, the solution of the strong form is equivalent to the minimizer of the energy functional.  
- The weak form may be defined as

$$
G(u, v) = \mathcal{A}(u, v) - \mathcal{L}(v) = 0 \quad \forall v \in \mathcal{U}_0
$$

with the two operators
$$
\begin{align}
\mathcal{A}(u, v) &= \int_\Omega \sigma_{ij}(u) v_{i,j} \, dV \\
\mathcal{L}(v) &= \int_\Omega \rho b_i v_i \, dV + \int_{\partial \Omega_N} \hat t_i v_i \, dS
\end{align}
$$

## Dynamics

- Inertial effects (dynamic response) gain importance when the mass of the body affects the dynamics
- Create an initial boundary value problem
- Static loading is generally time-independent
- Dynamic loads are time dependent (inertial effects are not negligible)
- Quasistatic: loading is time dependent but is sufficiently slow such that inertial effects can be safely ignored

**Strong Form of the Dynamic Mechanical Boundary Value Problem**
$$
\begin{align}
\sigma_{ij,j} + \rho b_i &= \rho \ddot u_i &\qquad \text{in } \Omega \\
u_i(\mathbf x, t) &= \hat u_i(\mathbf x, t) &\qquad \text{on } \partial \Omega_D \\
\sigma_{ij} n_j(\mathbf x, t) &= \hat t_i(\mathbf x, t) &\qquad \text{on } \partial \Omega_N
\end{align}
$$

- with suitable initial conditions
- the displacement field $u_i(\mathbf x, t)$ varies with position $\mathbf x \in \Omega$ and time $t \in \mathbb{R}$
- Required sufficient differentiability of the solution $u_i$ in space and time





**Variational Structure**

- Strong form of the dynamic mechanical boundary value problem can be cast into a variational structure by making use of the so called action principle
- We define the action of a deforming and moving body as the integral of the Lagrangian $L$ over time

$$
\mathcal{A}[u] = \int_{t_1}^{t_2} L[u] dt \qquad \text{with} \qquad L[u] = T[u] - I[u]
$$

being the Lagrangian.
$I$ is the potential energy functional and $T$ is the kinetic energy functional.
For a variational material model with energy density $W$, and assuming constant body forces, $\rho \mathbf b$ and surface tractions $\hat t$, we have

$$
\mathcal{A}[u] =
\int_{t_1}^{t_2} \left[\int_\Omega \frac{1}{2} \rho ||\dot u|^2 - W(\epsilon) dV 
+
\int_\Omega \rho \mathbf b \cdot \mathbf u dV
+
\int_{\partial \Omega_N} \hat t \cdot u dS \right] dt
$$

- Action principle (principle of stationary action): The solution $\mathbf{u}(\mathbf x, t)$ renders the action $\mathcal{A}$ stationary, under suitable initial conditions
- Condition is analogous to rendering the total potential energy $I$ stationary in the static case, but now we need to consider the kinetic energy $T$ as well

- By taking the first variation of the action $\mathcal{A}$ with respect to the displacement field $u$ (using the divergence on the stress term), we obtain the stationary condition

$$
\delta \mathcal{A} = 
\int_{t_1}^{t_2} \left[
    \int_\Omega (\rho \dot u_i \delta \dot u_i - \sigma_{ij} \delta u_{i,j}) \, dV
    +
    \int_\Omega \rho b_i \delta u_i \, dV
    +
    \int_{\partial \Omega_N} \hat t_i \delta u_i \, dS
\right] dt
= 0
$$

- The dynamic weak form is thus given by
$$
\mathcal{G}[u, v] = \int_{t_1}^{t_2} \left[
    \int_\Omega (\rho \dot u_i \dot v_i + \sigma_{ij} v_{i,j}) \, dV
    -
    \int_\Omega \rho b_i v_i \, dV
    -
    \int_{\partial \Omega_N} \hat t_i v_i \, dS
\right] dt = 0
$$

- The weak from indicates the space of solutions, and it particularly let us conclude that
$$
\mathbf v \in \mathcal{U_0} = \{ \mathbf v \in H^1(\Omega) \times H^1(\mathbb{R}): \mathbf v = 0 \text{ on } \partial \Omega_D \text{ and at } t = t_1  \text{ or } t = t_2 \}
$$

- where the last two conditions on the test function $\mathbf v$ are due to the assumption of $\mathbf u(\mathbf x, t_1) = u_1(\mathbf x)$ at $t = t_1$ and $\mathbf u(\mathbf x, t_2) = u_2(\mathbf x)$ at $t = t_2$ (initial conditions)
- We need the solution to have square integrable first derivatives with respect to both time and space.

- The time derivate $\dot v$ can be removed by integration by parts in time and the boundary term vansihes due to the condition $\mathbf v = 0$ at $t = t_1$ and $t = t_2$. This leads to

$$
\mathcal{G}[u, v] = - \int_{t_1}^{t_2} \left[
    \int_\Omega (\rho \ddot u_i v_i + \sigma_{ij} v_{i,j}) \, dV
    -
    \int_\Omega \rho b_i v_i \, dV
    -
    \int_{\partial \Omega_N} \hat t_i v_i \, dS
\right] dt = 0
$$

## Time Integration Schemes

**Semi-Discretization**
- Since the dynamic problem displacement field depends on time, we here introduce a semi-discretization
- We discretized the solution in space but not yet in time
- This implies that we introduce shape functions of the interpolation in space, where the nodal unknowns $\mathbf u^a$ become functions of time:

$$
\mathbf{u}^h(\mathbf x, t) = \sum_{a=1}^{n} \mathbf{u}^a(t) N^a(\mathbf x)
\qquad \text{and} \qquad
\mathbf{v}^h(\mathbf x, t) = \sum_{a=1}^{n} \mathbf{v}^a(t) N^a(\mathbf x)
$$

- The shape functions $N^a(\mathbf x)$ can be the same as in the static case, but now the nodal unknowns $\mathbf u^a$ are functions of time

- The semi-discretization allows us to calculate time derivatives, e.g. the approximate velocity and acceleration fields as

$$
\mathbf{\dot{u}}^h(\mathbf x, t) = \sum_{a=1}^{n} \mathbf{\dot{u}}^a(t) N^a(\mathbf x)
\qquad \text{and} \qquad
\mathbf{\ddot{u}}^h(\mathbf x, t) = \sum_{a=1}^{n} \mathbf{\ddot{u}}^a(t) N^a(\mathbf x)
$$

Insertion into the weak form results in the semi-discrete weak form

$$
\mathcal{G}[\mathbf{u}^h, \mathbf{v}^h] = - \int_{t_1}^{t_2} 
    \sum_{b=1}^{n} v_i^b(t) \left[
        \sum_{a=1}^{n} \ddot u_i^a 
        \int_\Omega \rho N^a N^b \, dV
        +
        \int_\Omega \sigma_{ij} N^b_{,j} \, dV
        -
        \int_\Omega \rho b_i N^b \, dV
        -
        \int_{\partial \Omega_N} \hat t_i N^b \, dS
\right] dt = 0
$$

- which must hold for all $\mathbf{v}^b(t)$ histories that vanish at $t = t_1$ and $t = t_2$.
- We conclude that the term in square brackets must vanish for all $b=1, \ldots, n$ and $i= 1, \ldots, d$ ($n$ is the number of nodes and $d$ is the spatial dimension)
- These $n \times d$ equations are the semi-discrete equations of motion for the dynamic problem
- We collect all time-dependent nodal dofs in the vector $\mathbf{U}(t) = [\mathbf{u}^1(t), \ldots, \mathbf{u}^n(t)]^T$ and we can write the semi-discrete equations of motion in matrix form as
$$
\mathbf{M} \ddot{\mathbf{U}} + \mathbf{F}_{\text{int}}(\mathbf{U}) = \mathbf{F}_{\text{ext}}(t)
$$

with

$$
\begin{align}
\mathbf{M}_{ij}^{ab} &= \delta_{ij} \int_\Omega \rho N^a N^b \, dV \\
\mathbf{F}_{\text{int}, i}^{b} &= \int_\Omega \sigma_{ij} N^a_{,j} \, dV \\
\mathbf{F}_{\text{ext}, i}^{b} &= \int_\Omega \rho b_i N^b \, dV + \int_{\partial \Omega_N} \hat t_i N^b \, dS
\end{align}
$$

- Matrix $\mathbf{M}$ is the consistent mass mastrix
- $\mathbf{F}_{\text{ext}} = \mathbf{F}_{\text{ext}}(t)$ is the external force vector, which is time dependent due to the time dependence of the body forces and surface tractions

## Time Integration Schemes

- Finite difference schemes are employed to obtain solutions in a time-discretized fashion
- This converts the previously introduced semi-discretization of the fields of interest into a proper discretization in both space and time.
- Finite differences are a conventient technique when the discretization is applied to a regular grid with uncomplicated boundary conditions
- Example: choosing constant time increments with known initial conditions
- We define a constant time increment $\Delta t > 0$ and seek solutions at discrete time intervals:
$$
\Delta t = t^{\alpha + 1} - t^{\alpha}, \qquad
\mathbf{u}^a(t^\alpha) = \mathbf{u}^{a, \alpha} \qquad
\mathbf{U}^\alpha = \mathbf{U}(t^\alpha) \qquad
\forall \alpha \geq 0
$$
- We must choose between explicit and implicit integration schemes

### Implicit Time Integration Schemes
- Implicit schemes require solving a generally nonlinear system of equations for $U^{\alpha + 1}$ at each time step,

#### Newmark's method
- Most popular integration scheme for mechanical problems
- Motivation: The solution vector $\mathbf{U}(t)$ at some time $t \in (t^\alpha, t^{\alpha + 1})$ can be obtained from knowledge of $\mathbf{U}^\alpha$ and $\dot{\mathbf{U}}(t)$ within that interval by integration
$$
\mathbf{U}(t) = \mathbf{U}^\alpha + \int_{t^\alpha}^{t} \dot{\mathbf{U}}(\tau) d\tau
$$
- Analogously, we know that the velocity $\dot{\mathbf{U}}(t)$ can be obtained from knowledge of $\dot{\mathbf{U}}^\alpha$ and $\ddot{\mathbf{U}}(t)$ within that interval by integration
$$
\dot{\mathbf{U}}(t) = \dot{\mathbf{U}}^\alpha + \int_{t^\alpha}^{t} \ddot{\mathbf{U}}(\tau) d\tau
$$

- In practice we do not know the velocity $\dot{\mathbf{U}}(t)$ and acceleration $\ddot{\mathbf{U}}(t)$ within the interval
- But we may define those at the discrete time steps $t^\alpha$ and $t^{\alpha + 1}$
- Time integration schemes can be devised by introduction approximated for how the acceleration $\ddot{\mathbf{U}}(t)$ is defined during the interval $[t^\alpha, t^{\alpha + 1}]$

- **Average Acceleration Method**: The method assumes that the acceleration is piecewise constant and, between two time steps, assumes the average value of the acceleration at the two time steps
$$
\ddot{\mathbf{U}}(t) = \frac{\ddot{\mathbf{U}}^\alpha + \ddot{\mathbf{U}}^{\alpha + 1}}{2} = \text{constant} \qquad \forall t \in [t^\alpha, t^{\alpha + 1}]
$$
- Inserting this into the above equations for $\mathbf{U}(t)$ and $\dot{\mathbf{U}}(t)$ and evaluating at $t = t^{\alpha + 1}$ gives
$$
\begin{align}
\mathbf{U}^{\alpha + 1} &= \mathbf{U}^\alpha + \Delta t \dot{\mathbf{U}}^\alpha + \frac{\Delta t^2}{4} (\ddot{\mathbf{U}}^\alpha + \ddot{\mathbf{U}}^{\alpha + 1}) \\
\dot{\mathbf{U}}^{\alpha + 1} &= \dot{\mathbf{U}}^\alpha + \frac{(\Delta t)}{2} (\ddot{\mathbf{U}}^\alpha + \ddot{\mathbf{U}}^{\alpha + 1})
\end{align}
$$
- This is an implicit definition of the new velocity and displacement vectors at time $t^{\alpha + 1}$, since they involve the unknown acceleration $\ddot{\mathbf{U}}^{\alpha + 1}$ at time $t^{\alpha + 1}$.

**Generalized Newmark-$\beta$ method**
- The method generalized the average acceleration method by introducing two parameters $\beta$ and $\gamma$ that allow to control the numerical dissipation of the method
$$
\begin{align}
\mathbf{U}^{\alpha + 1} &= \mathbf{U}^\alpha + \Delta t \dot{\mathbf{U}}^\alpha + \frac{\Delta t^2}{2} ((1 - 2\beta) \ddot{\mathbf{U}}^\alpha + 2\beta \ddot{\mathbf{U}}^{\alpha + 1}) \\
\dot{\mathbf{U}}^{\alpha + 1} &= \dot{\mathbf{U}}^\alpha + \Delta t ((1 - \gamma) \ddot{\mathbf{U}}^\alpha + \gamma \ddot{\mathbf{U}}^{\alpha + 1})
\end{align}
$$

- With $\beta = 1/4$ and $\gamma = 1/2$, we recover the average acceleration method. This scheme is unconditionally stable and second-order accurate without numerical damping.
- With $\beta = 0$ and $\gamma = 1/2$, we recover the explicit central difference method.
- With $\beta = 1/6$ and $\gamma = 1/2$, we recover the linear acceleration method.

**Algorithmic Structure**
- At each time step, we solve for the unknown displacement vector $\mathbf{U}^{\alpha + 1}$ at time $t^{\alpha + 1}$ by solving the nonlinear system of equations given by the semi-discrete equations of motion using an iterative method for nonlinear equations, based on the information at the previous time step $t^\alpha$ and the external applied forces at time $t^{\alpha + 1}$.
- This implies we must store vectors $\mathbf{U}^\alpha$, $\dot{\mathbf{U}}^\alpha$, and $\ddot{\mathbf{U}}^\alpha$ after each converged time step
- Once the new solution $\mathbf{U}^{\alpha + 1}$ is obtained, we can update the velocity and acceleration vectors at time $t^{\alpha + 1}$ using the Newmark-$\beta$ method, store all three vectors for the next time step, and repeat the process until we reach the final time $t = t_2$.
