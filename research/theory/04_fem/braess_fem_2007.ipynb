## Introduction to Elasticity Theory

- Elasticity deals with the deformation of bodies under the influence of applied forces and resulting stresses and strains which result from the deformation.

### Kinematics

- We know a reference configuration $\bar \Omega$ for the considered body.
    - $\bar \Omega$ is the closure of a bounded open set $\Omega$
    - $\bar \Omega$  is the undeformed configuration of the body (subset of $\mathbb{R}^3$)

- Current state is given as a mapping: $\phi: \bar \Omega \to \mathbb{R}^3$
    - $\phi(x)$ is the position of  a point which was located at $x$ in the reference configuration
    - This is: $\phi = \text{id} + u$ where $u$ is the displacement field
    - Assume displacements are small, and we will neglect higher order terms in $u$

- Rigid body motions do not alter the stress in a body (this invariance must be preserved in the finite element results)

- Assume mapping $\phi$ is sufficiently smooth, $\phi$ represents a deformation if $\det \nabla \phi > 0$
    - $\nabla \phi$ is the deformation gradient, and
    - $\det \nabla \phi$ is the local volume change

- Subdomains with positive volume are mapped into subdomains with positive volume.

- Defoemations are injective mappings locally. The mapping induces:
$$ \phi(x+z) - \phi(x) = \nabla \phi(x) \cdot z + o(|z|) $$

- In terms of Euclidean distance,
$$ \|\phi(x+z) - \phi(x)\| = \|\nabla \phi(x) \cdot z\| + o(|z|) $$

- Thus the following matrix describes the transformation of the length elements
$$ C := \nabla \phi^T \cdot \nabla \phi $$
   - $C$ is the right Cauchy-Green deformation tensor
   - $C$ is symmetric

- The deviation of $C$ from the identity matrix describes the deformation of the body, and is given by
    - $E$ is the Green-Lagrange strain tensor
    - $E$ is symmetric
$$ E := \frac{1}{2} (C - I) $$

- Quadratic terms are neglected in the linearized theory, and we have
    - $\epsilon$ is the linearized strain tensor
    - $\epsilon$ is symmetric

$$ \epsilon_{ij} = \frac{1}{2} \left( \frac{\partial u_i}{\partial x_j} + \frac{\partial u_j}{\partial x_i} \right) $$

- Rigid body motion: $C(x) = I$ for all $x \in \bar \Omega$

### The Equilibrium Equations

- Interaction of the body with the outside world is described by:
    - Applied surface forces (forces distributed over the surface) $\to$ force due load on a bridge
    - Applied body forces (forces distributed over the volume) $\to$ force of gravity
- **Body Forces**: $f: \Omega \to \mathbb{R}^3$ results in a force $f \, dV$ acting on a volume element $dV$
- **Surface Forces**: $t: \Omega \times S^2 \to \mathbb{R}^3$ where $S^2$ is the unit sphere
    - Let $V$ be an arbitrary subdomain of $\Omega$ with boundary $\Gamma$
    - Let $dA$ be an area element of $\Gamma$ with outward normal $n$
    - The area element $dA$ contributes $t(x, n) \, dA$ to the force, which also depends on the direction of the normal vector $n$
    - Vector $t(x, n)$ is called the Cauchy stress vector

**Axiom of Static Equilibrium**:

Let B be a deformed body in equilibrium.
Then there exists a vector field $t$ such that in every subdomain $V$ of $B$, the volume forces f and stresses $t$ satisfy
$$\begin{aligned}
\int_V f(x) \, dx + \int_{\partial V} t(x, n) \, ds = 0 \\
\int_V x \times f(x) \, dx + \int_{\partial V} x \times t(x, n) \, ds = 0
\end{aligned}$$

**Notation**
- $\mathbb M^3$: set of all $3 \times 3$ matrices
- $\mathbb M_+^3$: set of all $3 \times 3$ matrices with positive determinant
- $\mathbb O^3$: set of all $3 \times 3$ orthogonal matrices
- $\mathbb O_+^3 := \mathbb O^3 \cap \mathbb M_+^3$: set of all $3 \times 3$ orthogonal matrices with positive determinant
- $\mathbb S^3$: set of all $3 \times 3$ symmetric matrices
- $\mathbb S_+^3$: set of all positive definite $3 \times 3$ symmetric matrices


**Cauchy's Theorem**:

Let $t(\cdot, n) \in C^1(B, \mathbb{R}^3), t(x, \cdot) \in C^0(S^2, \mathbb{R}^3)$ and $f \in C(B, \mathbb{R}^3)$ be in equilibrium.
Then there exists a symmetric tensor field $T \in C^1(B, \mathbb{R}^{3 \times 3})$ such that
$$\begin{aligned}
t(x, n) = T(x) n \qquad &x \in B, n \in S^2 \\
\text{div} \, T(x) + f(x) = 0 \qquad &x \in B \\
T(x) = T^\top(x) \qquad &x \in B
\end{aligned}$$
The tensor $T$ is called the Cauchy stress tensor.

- Key assertion is the representability of the stress vector $t$ in terms of a stress tensor $T$.
- Using the Gauss integral theorem, the equilibrium equations can be written as
$$
\int_V f(x) \, dx + \int_{\partial V} T(x) n \, ds = \int_V \left[f(x) + \text{div} \, T(x) \right] dx = 0
$$


### The Piola Transform
- Equilibrium equations are fomulated in terms of coordinated of the deformed body $B$ (as did Euler)
- Useful to transform the variables to the reference configuration since these coordinates have to be computed in the first place
- Subscript $R$ denotes the reference configuration: $x = \phi (x_R)$

- Volume element is given by $dx = \det (\nabla \phi) dx_R$
- Forces are proportional to the density and densities are transformed according to conservation of mass:
$$ \rho(x) dx = \rho_R(x_R) dx_R \implies \rho(x) = \det (\nabla \phi^{-1}) \rho_R(x_R) $$
- Consequently, the body forces are transformed according to
$$ f(x) = \det (\nabla \phi^{-1}) f_R(x_R) $$

- Transformation of the stress tensor is more complicated. In terms of reference configuration we have
$$ \text{div}_R T_R + f_R = 0 $$
where
$$ T_R := \det (\nabla \phi) T  (\nabla \phi)^{-\top} $$
- In contrast to the Cauchy stress tensor $T$, the Piola-Kirchhoff stress tensor $T_R$ is not symmetric.
- To achieve symmetry, we introduce the second Piola-Kirchhoff stress tensor
$$ \Sigma_R := \det (\nabla \phi) (\nabla \phi)^{-1} T  (\nabla \phi)^{-\top} $$
- Clearly: $\Sigma_R = (\nabla \phi)^{-1} T_R$
- Differences between stress tensors can be neglected for small deformation gradients

### Constitutive Equations
- Important problem to find the deformation of a body and the associated stress corresponding to given external forces
Equilibrium equations does not determine the 6 component of the symmetric stress tensor.
- Missing equations arise from constitutive equations (they express how the deformation deforms on properties of the material)

**Definition**: A material is called elastic if there is a mapping  
$$\hat T: \mathbb M_+^3 \to \mathbb{R}^{3 \times 3}$$
such that for every deformation $\phi$ the stress is given by

$$ T(x) = \hat T(\nabla \phi(x_R)) $$

The mapping is called the response function of the Cauchy stress, and is called the constitutive equation.

- Constitutive equation: stress depends on the displacement in a local way.
- We introduce the response function for the second Piola-Kirchhoff stress
$$ \hat \Sigma(F) := \det (F) F^{-1} \hat T(F) F^{-\top} $$
where $F = \nabla \phi(x)$ is the deformation gradient.
- Homogenous materials: $\hat T$ does not depend on $x$
- Cauchy stress vector is independent of the choice of coordinates: Frame-indifferent material is also called objective
$$Qt(x,n) = t(Qx, Qn) \qquad \forall Q \in \mathbb O_+^3$$

**Theorem**: Suppose the material frame-indifference holds. Then for every orthogonal transformation $Q \in \mathbb O_+^3$
$$
\hat T(QF) = Q \hat T(F) Q^\top
$$

Moreover, there exists a mapping $\tilde \Sigma: \mathbb S_>^3 \to \mathbb S^3$ such that
$$
\hat \Sigma(F) = \tilde \Sigma(F^\top F)
$$
i.e, the second Piola-Kirchhoff stress depends only on the Green-Lagrange strain tensor $C=F^\top F$.

**Definition**
A material is called isotropic provided
$$ \hat T(F) = \hat T(QF) \qquad \forall Q \in \mathbb O_+^3 $$


**Rivelin-Ericksen Theorem**: A response function $\hat T: M_+^3 \to \mathbb{R}^{3 \times 3}$ is objective and isotropic if and only if it has the form $\hat T(F) = \bar T (F F^\top)$,and,
$$
\begin{aligned}
\bar T: \mathbb S_+^3 &\to \mathbb S^3 \\
\bar T(B) &= \beta_0(l_B) I + \beta_1(l_B) B + \beta_2(l_B) B^2
\end{aligned}
$$
Here, $\beta_0, \beta_1, \beta_2$ are functions of the invariants of $B$, and $l_B$ is the vector of invariants of $B$.

**Remarks**
- In special case where $F F^\top = c I$ for some $c > 0$, the response function reduces to $\hat T(F) = \alpha(c) I$ for some function $\alpha: \mathbb R_+ \to \mathbb R$.
- The stress has then the character of a pure pressure

**Corollary**: $\Sigma(\nabla \phi) = \tilde \Sigma(C)$ for an isotropic and objective material, where $C = \nabla \phi^\top \nabla \phi$ is the Green-Lagrange strain tensor and 
$$
\tilde \Sigma(C) = \gamma_0 I + \gamma_1 C + \gamma_2 C^2
$$
where $\gamma_0, \gamma_1, \gamma_2$ are functions of the invariants of $l_C$.
Thus, the second Piola-Kirchhoff stress depends only on the Green-Lagrange strain tensor.

### Linear Material Laws

- stress-strain relationship can be described in terms of 2 parameters in the neighborhood of a strain-free reference configuration
- Setting $C = I + 2E$ we obtain
$$ \tilde \Sigma(I + 2E) = \gamma_0(E)I + \gamma_1(E)E + \gamma_2(E)E^2 $$

**Theorem**:
Suppose that in addition to the hypotheses of Corollary 1, $\gamma_i$ are differentiable functions of the invariants of $E$.
Then there exist numbers $\pi, \lambda, \mu$ with
$$
\tilde \Sigma(I + 2E) = -\pi I + \lambda \text{trace}(E) I + 2 \mu E + o(|E|) \qquad \text{as } E \to 0
$$

- Situation $C=I$ corresponds to an unstressed condition, and $\pi=0$
- Other 2 constants are called Lamé constants
- If higher order terms are neglected, we obtain the linear material law of Hooke:
$$ \tilde \Sigma(I + 2E) = \lambda \text{trace}(E) I + 2 \mu E $$

- A material which satisfies the linear material law and not just for small strains is called a St. Venant-Kirchhoff material.

- $\text{trace}E= \text{div} u$ is the volumetric strain, $\lambda$ describes the stress due to change in the density of the material
- $\mu$ describes the stress due to change in the shape of the material, and is called the shear modulus

- Relation between Lamé constants and Young's modulus $E$ and Poisson's ratio $\nu$:
$$\begin{aligned}
E &= \frac{\mu (3 \lambda + 2 \mu)}{\lambda + \mu} \\
\nu &= \frac{\lambda}{2 (\lambda + \mu)}
\end{aligned}$$

- Poisson's ratio $\nu$  describes the influence of the stresses on the displacement in the orthogonal direction
- Deformations, stresses and strains are defined by the kinematics, the equilibrium equations, and the constitutive equations
- In principle, only the equilibrium equations (for the Cauchy stress tensor) are linear
- Small Deformations: replace the strain $E$ by the linearization $\epsilon$, it suffices to work with the so called geometrically linear theory

## Hyperelastic Materials


**Definition**: An elastic material is called hyperelastic if there exists an energy functional
$\hat W: \Omega \times \mathbb M_+^3 \to \mathbb R$ such that
$$ \hat T(x, F) = \frac{\partial \hat W}{\partial F} (x, F) \qquad \text{ for } x \in \Omega, F \in \mathbb M_+^3 $$

- There is a variational formulation corresponding to the boundary value problem for hyperelastic materials, provided that the vector fields $f$ and $g$ can be represented as gradients of potential functions: $f = \text{grad} \mathcal F$ and $g = \text{grad} \mathcal G$ for some $\mathcal F, \mathcal G: \Omega \to \mathbb R$.
- The solutions of the boundary value problem for hyperelastic materials are stationary points of the total energy functional

$$
I(\psi) = \int_\Omega \left[ \hat W(x, \nabla \psi(x)) - \mathcal F(x) \right] \, dx + \int_{\Gamma_1} \mathcal G(\psi(x)) \, dx
$$

- As deformations we admit functions $\psi$ which satisfy Dirichlet boundary conditions on $\Gamma_0$ along with the local injectivity condition $\det \nabla \psi > 0$.

- Equation above refers to the variational formulation for the displacements

- For an objective material, the energy density $\hat W$ is a function of only $C = F^\top F$ and we have
$$
\hat W(x, F) = \tilde W(x, F^\top F)
$$

and 

$$
\tilde \Sigma(x, C) = 2 \frac{\partial \tilde W (x, C)}{\partial C} \qquad \forall \,  C \in \mathbb S_+^3
$$

- $\tilde W$ depends only on the principal inavariants of $C$, i.e., $\tilde W(x, C) = \dot W(x, l_C)$ where $l_C$ is the vector of invariants of $C$.
- Analogously, for isotropic materials we have
$$
\tilde W(x, F) = \dot W(x, FQ) \qquad \forall Q \in \mathbb O_+^3, F \in \mathbb M_+^3
$$

**Neo-Hookean Material**:

$$
\tilde W(x, C) = \frac{1}{2} \mu\left[ \text{trace}(C - I) + \frac{2}{\beta} \{ (\text{det} C)^{-\beta /2} - 1 \} \right]
$$

where $\beta = \frac{2 \nu}{1 - 2 \nu}$ and $\mu$ is the shear modulus.
