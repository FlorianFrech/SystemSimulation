# Actuator

## Description

We select the drive system [ILM-E50x08](https://www.tq-group.com/filedownloads/files/products/robodrive/data-sheets/de/DRVA_DB_Servo-Kits_ILM-E_DE_Rev0106_Web.pdf) which is a PMSM (Permanent Magnet Synchronous Motor) (a.k.a. brushless DC motor).

We assume it this brushless DC motor is used in star parallel configuration. This means, that the three phases of the motor are connected a common neutral. Further, multiple windings are set in parallel to increase the current capacity. 

The system-level approximation of the PMSM by a DC motor model is justified by the fact that the control of the PMSM is done by a field-oriented control (FOC) which decouples the torque and flux control. This allows us to model the torque control as a simple DC motor model. We are not interested in the modeling of the phase currents explicitly.

The current limitation is imposed by the motor drive electronics (current controller / power stage), thermal limits (continuous current), and the power supply limits (peak current). The continuous current is the maximum current that the motor can handle without overheating, while the peak current is the maximum current that the motor can handle for a short period of time (e.g., during acceleration or deceleration).

## Modeling Assumptions

A BLDC / PMSM is usually operated with an inverter and a field-oriented control (FOC) which decouples the torque and flux control.
The approximation by a simpler DC motor model is only valid under the following assumptions:
- The control signal lead directly to a torque-producing current
- The machine is not dominated by current and voltage limits in the operating range of interest

In the **star parallel configuration**, the current **I** in the model represents the **replacement current** to produce the actual torque but not the actual current in the windings.

## Simulation Requirements

For our dirve selection, we have the following requirements on the electrical time constant:

$$\tau_e = \frac{L}{R} = \frac{121 \, \mu H}{0.151 \, \Omega} \approx 0.8 \, ms$$

Thus, we require an adaptive time step solver with a maximum time step of $0.8 \, ms$ to accurately capture the electrical dynamics of the system.
