```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, FancyArrowPatch, Arc
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.set_aspect("equal"); ax.axis("off")

# pivot + rod (example)
ax.add_patch(Circle((0,0), 0.15, fill=False, lw=2))
ax.plot([0, 3], [0, -2], lw=6, solid_capstyle="round")  # rod

# blocks (A/D, PC, D/A, Drive)
for i, label in enumerate(["A/D","PC","D/A","Drive"]):
    ax.add_patch(Rectangle((-1+1.4*i, -3.2), 1.1, 0.6, fill=False, lw=2))
    ax.text(-1+1.4*i+0.55, -2.9, label, ha="center", va="center")

fig.savefig("controlled_pendulum_setup.svg", bbox_inches="tight")
```
