```python
import sys
import os
import subprocess
import pathlib
from pathlib import Path
from IPython.display import SVG, display

src_dir = Path.cwd() / 'src'
src_files = list(src_dir.rglob('*.puml'))

for i, puml_file in enumerate(src_files):
    print(f'Processing file {i + 1}/{len(src_files)}: {puml_file}')
    outdir = Path.cwd() / 'out'
    output_file = outdir / puml_file.with_suffix('.svg').name
    subprocess.run(['plantuml', '-tsvg', str(puml_file), '-o', str(outdir)], check=True)
```
