```python
import sys
import os
import subprocess
import pathlib
from pathlib import Path
from IPython.display import SVG, display

src_dir = Path.cwd() / 'src'
src_files = list(src_dir.rglob('*.puml'))

for puml_file in src_files:
    outdir = Path.cwd() / 'out'
    output_file = outdir / puml_file.with_suffix('.svg').name
    subprocess.run(['plantuml', '-tsvg', str(puml_file), '-o', str(outdir)], check=True)
```

```python
# Show the generated files
for svg_file in outdir.rglob('*.svg'):
    print(f'Generated: {svg_file}')
```

```python
# Display the generated SVG files in a Jupyter cell
for svg_file in outdir.rglob('*.svg'):
    display(SVG(filename=str(svg_file)))
```
