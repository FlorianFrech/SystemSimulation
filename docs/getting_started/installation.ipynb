# Installation


## Requirements

SysSimX requires **Python 3.11 or later**. We recommend using a virtual environment.

**Core Dependencies** (installed automatically):

- `numpy` - numerical computing
- `scipy` - scientific computing utilities
- `pint` - physical units handling
- `networkx` - graph algorithms
- `matplotlib` - plotting (used in tutorials and demos)
- `ipywidgets` and `traitlets` - UI components

**Optional Extras** (install only what you need):

- `fmu` - FMU co-simulation support (`fmpy`, `OMPython`)
- `fem` - FEM support (`ngsolve`, `netgen`)
- `opensim` - OpenSim musculoskeletal models (`opensim`)
- `viz` - system graph visualization (`graphviz`, `pydot`)
- `dev` - development tools (tests, linting, docs, notebooks)
- `all` - all optional features
- `full` - all optional features plus dev tools

```{note}
Some extras require external system tools. For example, OpenModelica is needed for
OMPython usage, and Graphviz must be installed at the system level for graph rendering.
```


## Installation Methods


### From PyPI (Recommended)


````python
pip install syssimx
````

Install optional extras as needed:


````python
pip install syssimx[fmu]
pip install syssimx[viz]
pip install syssimx[fem]
pip install syssimx[opensim]

# All optional features
pip install syssimx[all]

# Full development setup
pip install syssimx[full]
````

### From Source

If you are developing SysSimX or want the latest changes, install from source:


````python
git clone https://github.com/FlorianFrech/SystemSimulation.git
cd SystemSimulation
pip install -e .

# With optional extras
pip install -e .[fmu,viz]
````
