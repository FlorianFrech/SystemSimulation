# Installation


## Basic Setup

1) We are using a single virtual environment for all notebooks. This ensures that all dependencies are met and compatible with each other.

2) We install jupyter + ipykernel inside that virtual environment.

3) We register the virtual environment as a jupyter kernel, so that it can be selected in jupyter notebooks.

4) We start Jupyter from that virtual environment or select that kernel in VS Code.

## Setting up a fresh Environment with Conda

We refer to the [Anaconda documentation](https://www.anaconda.com/docs/getting-started/miniconda/install#windows-installation) for installing *Miniconda* or *Anaconda* on your system.

In a terminal (e.g., Anaconda Prompt, PowerShell, or bash), run the following commands:

```bash
conda create -n env_syssimx python=3.12
conda activate env_syssimx
conda install jupyter ipykernel

 conda create -n env_syssimx --override-channels -c conda-forge python=3.12 "numpy>=2.1.3,<2.7" scipy matplotlib jupyterlab ipykernel pip -y


````python
conda --version
````

## Installing SysSimX

Now, you can install SysSimX using pip:

**Development Version:**
```bash
python -m pip install -e .[full]
```

**Stable Version:**
```bash
python -m pip install syssimx
```

## Installing Graphviz

We need to ensure that the Graphviz executables are in our system's PATH.
On Windows, you can download the Graphviz installer from the [Graphviz website](https://graphviz.org/download/). During installation, make sure to select the option to add Graphviz to your system PATH.

```bash
conda install -c conda-forge graphviz
```

## Installing OpenModelica

We refer to the [OpenModelica installation guide](https://github.com/OpenModelica/OMPython) for detailed instructions on how to install OpenModelica's `OMPython` on your system.

```bash

```

## Requirements

**Core Dependencies** (installed automatically):

- `numpy` - numerical computing
- `scipy` - scientific computing utilities
- `pint` - physical units handling
- `networkx` - graph algorithms
- `matplotlib` - plotting (used in tutorials and demos)
- `ipywidgets` and `traitlets` - UI components

**Optional Extras** (install only what you need):

- `fmu` - FMU co-simulation support (`fmpy`, `OMPython`)
- `fem` - FEM support (`ngsolve`, `netgen`)
- `opensim` - OpenSim musculoskeletal models (`opensim`)
- `viz` - system graph visualization (`graphviz`, `pydot`)
- `dev` - development tools (tests, linting, docs, notebooks)
- `all` - all optional features
- `full` - all optional features plus dev tools
